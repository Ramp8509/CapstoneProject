# How Generative AI was used in the project

Generative AI was used as an intelligent assistant to support the development of the machine learning pipeline, improve documentation, and generate explanatory content. All code generated by the AI was reviewed, modified, tested, and validated before being incorporated into the final project.


**Benefits of Using Generative AI**

Reduced development time.
Assisted in debugging Python errors.
Improved understanding of machine learning concepts.
Helped document the project professionally.
Suggested best practices for fraud detection.

**Ethical Use Statement**

Generative AI was used responsibly as an educational assistant to support coding, documentation, and concept clarification. All generated code was reviewed, modified where necessary, executed locally, and validated by the author. Final design decisions, testing, interpretation of results, and project conclusions were made by the author.
